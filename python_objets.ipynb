```python
# what is objects in python:
# objects are instances of classes in python. They are created from a class definition and can have attributes and methods. Objects can represent real-world entities or abstract concepts, and they can interact with each other through their methods and attributes. In python, everything is an object, including functions, modules, and even classes themselves.
# objects can be string, integer, list, dictionary, etc. They can also be user-defined classes that we create ourselves. Objects can have properties (attributes) and behaviors (methods) that define their characteristics and actions. For example, a car object may have attributes like color, make, and model, and methods like start_engine() and stop_engine().
```

```python
# what is object reference in python:
# object reference in python refers to the way that variables in python point to objects in memory. When you create a variable and assign it a value, that variable is actually a reference to an object in memory. This means that when you assign one variable to another, you are not creating a new
```

```python
x = 2
# x is a reference to the integer object 2 in memory. If we assign y to x, then y will also be a reference to the same integer object 2 in memory.
y = x
# Now both x and y are references to the same integer object 2 in memory. If we change the value of x, it will not affect y because they are both references to the same
# object in memory. However, if we change the value of the integer object that x and y are referencing, then both x and y will reflect that change because they are both referencing the same object in memory.
```

```python
# Difference between is and == in python:
# the "is" checks if two variables point to the same objects
# the "==" checks if the values of the objects are the same

name1 = "sevil"
name2 = "sevil"
print(name1 == name2)  # This will print True because the values of name1 and name2 are the same
print(name1 is name2)  # This will print True because name1 and name2 are referencing the same string object "sevil" in memory. In python, small strings are intern
```

```python
# If you want to create a separate object, you need to use methods like copy() or deepcopy() for complex objects.
# For example:
import copy
list1 = [1, 2, 3]
list2 = copy.copy(list1)  # This creates a shallow copy of list1, so list2 is a new object in memory.
print(list1 == list2)  # This will print True because the values of list1 and list2 are the same
print(list1 is list2)  # This will print False because list1 and list2 are different objects in memory  
```
