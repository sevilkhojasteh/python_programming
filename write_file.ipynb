```python
# to write we can use 'w', or 'a'
file_path = 'example.txt'
f = open(file_path, 'w')
```

```python
f.write('Hello, World!\n')
```

```python
f.flush()
```

```python
f.write("continue")
```

```python
# When you want to add a text to a file

with open(file_path, 'a') as f:
    f.write('This is append mode')
```

```python
with open('output.txs', 'w') as f:
    lines =[
        'Hello',
        'Next line',
        'continue'
    ]
    f.writelines(lines)
```

```python
# 'w' over writes the files
# If you want to be cautious you can use 'x'

with open('output.txs', 'x') as f:
    f.write('hello')
```

```python
import datetime

# Function to log message
def log_message(log_file, message):
    with open(log_file, 'a') as f:
        f.write(message + '\n')

log_message('log.txt','This is a log message')
```
