```python
with open(file_path) as f:
    content = f.read()

    print(content)
```

```python
with open(file_path, 'r') as f:
    line = f.readline()
    while line:
        print(line, end='')
        line = f.readline()
```

```python
with open(file_path, 'r') as f:
    lines = f.readlines()

for line in lines:
    print(line)
```
