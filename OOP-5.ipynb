**Single inheritence**

*Like parent and child*

Pet --> Dog

```python
class Vehicle:
    num_wheels = 4
    def __init__(self, model, year):
        self.model = model
        self.year = year
        
    
    def display_info(self):
        print(f"Model: {self.model}, Year: {self.year}")
        
```

```python
class Bike(Vehicle):
    def __init__(self, model, year):
        pass
```

```python
class Car(Vehicle):
    def __init__(self, model, year, color):
        super().__init__( model, year)  # Calls the __init__ of Vehicle
        self.color = color  # New attribute specific to Car

    def display_info(self):
        super().display_info()
        print(f'Color: {self.color}')
```

```python
obj = Car("Toyota", 2026, "Red")
```
