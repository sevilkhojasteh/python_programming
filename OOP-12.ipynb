```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __str__(self):
        return f"'{self.title}' by {self.author}"
```

```python
obj = Book("wuthering heights", "Emily Brunte")
```

```python
print(obj)
```

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __repr__(self):
        return f"{self.__class__.__name__}({self.title!r}, {self.author!r})"
```

```python
obj = Book("wuthering heights", "Emily Brunte")
```

```python
obj
```

```python
print(obj)
# If you don't have __str__ it falls back to the __repr__
```

```python
class Inventory:
    def __init__(self):
        self.items = {}

    def add_item(self, item, quantity):
        if item in self.items:
            self.items[item] += quantity
        else:
            self.items[item] = quantity

    def remove_item(self, item, quantity):
        if item in self.items:
            if self.items[item] <= quantity:
                del self.items[item]
            else:
                self.items[item] -= quantity

    def __len__(self):
        return sum(self.items.values())
```

```python
obj = Inventory()
obj.add_item("apple", 3)
obj.add_item("orange", 4)
obj.add_item("cherry", 6)
```

```python
obj.__len__()
```

```python
obj.remove_item("apple", 2)
obj.__len__()
```

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        if isinstance(other, Vector):
            return Vector(self.x + other.x, self.y + other.y)
        else:
            raise ValueError("Operand must be an instance of Vector")

    def __mul__(self, scalar):
        if isinstance(scalar, (float, int)):
            return Vector(self.x * scalar, self.y * scalar)
        else:
            raise ValueError("Scalar must be a float or an integer")

    def __eq__(self, other):
        if isinstance(other, Vector):
            return (self.x == other.x) and (self.y == other.y)
        
    def __ne__(self, other):
        if isinstance(other, Vector):
            return (self.x != other.x) or (self.y != other.y)

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"
```

```python
p1 = Vector(3, 4)
p2 = Vector(6, 8)
```

```python
p1+p2
```

```python
p1*2
```

```python
p1 == p2
```

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __lt__(self, other):
        return (self.x ** 2 + self.y ** 2) < (other.x ** 2 + other.y ** 2)

    def __le__(self, other):
        return (self.x ** 2 + self.y ** 2) <= (other.x ** 2 + other.y ** 2)

    def __gt__(self, other):
        return (self.x ** 2 + self.y ** 2) > (other.x ** 2 + other.y ** 2)

    def __ge__(self, other):
        return (self.x ** 2 + self.y ** 2) >= (other.x ** 2 + other.y ** 2)
```

```python
class Counter:
    def __init__(self, initial_count=0):
        self.count = initial_count

    def __call__(self, increment=1):
        self.count += increment
        return self.count
```

```python
counter = Counter()
```
