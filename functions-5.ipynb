```python
def sum_number_2(*args):
    return sum(args)
```

```python
sum_number_2(2, 3)
```

```python
sum_number_2(4, 5, 6, 7)
```

```python
def sum_number(a, b, *args):
    return sum(a, b, args)
```

```python
def concat_string(seperator, *args):
    return seperator.join(args)
```

```python
name = [
    'a',
    'b',
    'c'
]
```

```python
concat_string(' - ', 'a', 'b', 'c')
```

```python
concat_string(' - ', *name)
```

 **Argument Tuple Packing**

```python
def calculate_average(*score):
    if score:
        return sum(score) / len(score)
    else:
        return 0
```

```python
calculate_average(10, 20, 30, 40, 50)
```

**Argument Tuple Unpacking**

```python
def print_book_info(title, author, year):
    print(f"Title: {title}, author: {author}, Year: {year}")
```

```python
book_details = ('The great Gatsby', 'F. Scott', 1925)
```

```python
print_book_info(*book_details)
```

```python
book_list = ['To Kill a Mockingbird', 'Harper Lee', 1960]
```

```python
print_book_info(*book_list)
```

```python
book_set = {'1984', 'George Orwell', 1949}
```

```python
print_book_info(*book_set)
```

**Argument Dictionary Packing**

```python
def introduce_yourself(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
introduce_yourself(name="Alice", age=30, profession="Engineer")
```

**Argument Dictionary Unpacking**

```python
def display_info(name, age, profession):
    print(f"Name: {name}, Age: {age}, Profession: {profession}")
```

```python
person_info = {'name': 'Alice', 'age': 30, 'profession': 'Engineer'}
```

```python
display_info(**person_info)
```

```python
def plan_session(*args):
    print(f"Topics planned for this session: {list(args)}")
```

```python
topics = ['Python Basics', 'Data Types Fundamentals', 'Functions']
```

```python
plan_session(*topics)
```

```python
def session_details(**kwargs):
    print("Session Details: ")
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
details = {
    'Name': 'Python Development Basics',
    'Attendees': 40,
    'Room': 105
}
```

```python
session_details(**details)
```
