```python
sorted([1, 92, 83], key= lambda x: x -10 * (x // 10))
```

```python
list(filter(
    lambda x: x % 2 ==0,
    [1, 2, 3, 4, 5]
))
```

```python
list(map(lambda x: x + 1, [1, 2, 3]))
```

```python
names = ['Alice', 'Bob', 'Charlie']
ages = [25, 30, 35]
list(zip(names, ages))
```

```python
for first, age in zip(names, ages):
    print(first, age)
```
