```python
class SmartGridConnection:
    def __init__(self, sensor_id):
        self.sensor_id = sensor_id

    def __enter__(self):
        # This runs the moment you type 'with'
        print(f"Opening secure connection to sensor {self.sensor_id}...")
        self.connected = True
        return self # This becomes the variable after 'as'

    def __exit__(self, exc_type, exc_val, traceback):
        # This runs the moment the 'with' block ends or crashes.
        # The 'exc' variables automatically capture any errors that happened!
        print(f"Closing connection to Sensor {self.sensor_id}.")
        self.connected = False
        
        if exc_type is not None:
            print(f"ALERT: Handled a crash safely: {exc_val}")
        
        # Returning True tells Python "I handled the error, don't crash the program"
        return True 


with SmartGridConnection("GRID_001") as grid:
    print("Downloading energy data...")

    raise ConnectionError("Power spike detected!")

print("Program finished successfully!")      
```
