```python
from dataclasses import dataclass

# Add this "decorator" above your class
@dataclass
class User:
    username: str
    email: str
    age: str

player = User("Sevil", "sevil@exmaple.com", 24)


# It also automatically prints nicely!
print(player)
```

```python
import asyncio

async def download_file():
    print("Started downloading...")
    # asyncio.sleep simulates a slow 2-second download
    await asyncio.sleep(2) 
    print("Download finished!")

async def main():
    # Run the download function
    await download_file()

# You must use this to start an async program
asyncio.run(main())
```
