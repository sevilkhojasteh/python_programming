```python
import asyncio

async def download_file():
    print("Started downloading...")
    # asyncio.sleep simulates a slow 2-second download
    await asyncio.sleep(2) 
    print("Download finished!")

async def main():
    # Run the download function
    await download_file()

# You must use this to start an async program
asyncio.run(main())
```

```python
# Python Dataclass:

# The Old Way
class Player:
    def __init__(self, name: str, level: int, health: float):
        self.name = name
        self.level = level
        self.health = health

    # If we don't write this, printing the player looks ugly: <__main__.Player object at 0x10a2...>
    def __repr__(self):
        return f"Player(name='{self.name}', level={self.level}, health={self.health})"

    # If we don't write this, we can't do: player1 == player2
    def __eq__(self, other):
        if not isinstance(other, Player):
            return False
        return self.name == other.name and self.level == other.level and self.health == other.health

p1 = Player("Sevil", 10, 100.0)
print(p1)
```

The 3 Hidden Superpowers

When you type @dataclass, Python secretly injects three magical methods into your class:

__init__: It automatically assigns your variables so you can create the object: Player("Sevil", 10, 100.0).

__repr__: It automatically formats the object so it prints beautifully to the console.

__eq__: It automatically allows you to compare two objects. If p1 and p2 have the exact same stats, p1 == p2 will return True.

```python
from dataclasses import dataclass

@dataclass
class Player:
    name: str
    level: int
    health: float

p1 = Player("Sevil", 10, 100.0)
print(p1)
```

```python
@dataclass
class Player:
    name: str
    level: int = 1         # Automatically starts at level 1
    health: float = 100.0  # Automatically starts with 100 health
    is_alive: bool = True

# Now I only HAVE to provide the name!
new_player = Player("Sevil")
print(new_player)
# Output: Player(name='Sevil', level=1, health=100.0, is_alive=True)
```

```python
from dataclasses import dataclass, field

@dataclass
class Player:
    name: str
    # Correct way to give every new player their own empty list:
    inventory: list = field(default_factory=list) 

p1 = Player("Sevil")
p1.inventory.append("Sword")
```

```python
@dataclass(frozen=True)
class DatabaseConfig:
    host: str = "127.0.0.1"
    port: int = 5432

db = DatabaseConfig()

# If you try to change it later:
db.port = 8080 
# ERROR! FrozenInstanceError: cannot assign to field 'port'
```
