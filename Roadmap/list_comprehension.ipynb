```python
# These are the shortcuts for one line lists

# [ <what_to_put_in_the_new_list>  for  <item>  in  <old_list> ]
numbers = [1, 2, 3, 4, 5]
squared = [n*n for n in numbers]
squared
```

```python
# [ <expression>  for  <item>  in  <old_list>  if  <condition> ]
numbers = [1, 2, 3, 4, 5, 6]

evens = [n for n in numbers if n % 2 == 0]
evens
```

```python
# [ <value_if_true> else <value_if_false>  for  <item>  in  <old_list> ]
scores = [45, 90, 82, 20, 30, 75]

grades = ['Pass' if score >= 50 else 'Fail' for score in scores]
grades
```

```python
nums = [2, 4, 6, 8]
multiplied = [n*10 for n in nums]
multiplied
```

```python
words = ["Apple", "Banana", "Cherry", "Date"]
new_list = [n[0] for n in words]
new_list
```

```python
balances = [100, -50, 120, -10, 300]
positives = [n for n in balances if n > 0]
positives
```

```python
sentence = "hello world"
vowels = 'aeoiu'
my_vowel = [vowel for vowel in sentence if vowel in vowels]
my_vowel
```

```python
temps = [95, 40, 105, 20, 85]
tempreture = ['Hot' if temp > 80 else 'Cold' for temp in temps]
tempreture
```
