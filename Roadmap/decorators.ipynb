```python
def outer_function(msg):
    message = msg

    def inner_function():
        print(message)
    return inner_function

```

```python
x = outer_function('Hi')
```

```python
x()
```

```python
def my_decorator(func):
    def wrapper():
        print("--------before-------")
        func()
        print("--------after--------")

    return wrapper

```

```python
@my_decorator
def say_hello():
    print("Hello!")

```

```python
say_hello()
```

```python
def universal_decorator(func):
    def wrapper(*args, **kwargs):
        print(f"Calling function: {func.__name__}")
        result = func(*args, **kwargs)
        print("Finished execusion")
        return result
    
    return wrapper
```

```python
@universal_decorator
def add(a, b):
    return a + b
```

```python
print(add(3, 4))
```

```python
import time

def timer_decorator(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        print(f"function {func.__name__} took {end_time - start_time:.4f} seconds")
        return result
    return wrapper
```

```python
@timer_decorator
def heavy_computation():
    sum([i**2 for i in range(100000000)])
```

```python
heavy_computation()
```

```python
def add_salutation(func):
    def wrapper(name):
        print("Good Day!")
        func(name)
        print("Goodbye!")

    return wrapper

@add_salutation
def say_name(name):
    print(f"My name is {name}")

say_name("Sevil")
```

```python
def enforce_types(func):
    def wrapper(a, b):
        if isinstance(a, int) and isinstance(b, int):
            return func(a, b)
        else:
            print("Error: Invalid argument type")

    return wrapper

@enforce_types
def add(a, b):
    return a + b

add(3, 4)
```

```python
add("3", 4)
```

```python
def cach_result(func):
    my_dict = {}
    def wrapper(n):
        if n not in my_dict:
            my_dict[n] = func(n)
        return my_dict[n]
    return wrapper

@cach_result
def expensive_calculation(n):
    print(f"Calculating {n} ...")
    return n * n
```

```python
expensive_calculation(5)
```

```python
expensive_calculation(5)
```
