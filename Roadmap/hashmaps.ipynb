If the set is your go-to tool for finding unique items instantly, the Hashmap is your ultimate tool for finding linked data instantly.

In Python, a hashmap is simply a Dictionary (dict). It is the single most important data structure for optimizing code and solving complex algorithmic problems.

1. The Core Concept: The Coat CheckThink of an array or linked list like walking through a massive parking lot looking for your car. You have to check every single spot ($O(n)$ time).A Hashmap is like an exclusive VIP Coat Check:You hand the attendant your coat (the Value) and tell them your name (the Key).The attendant performs a quick mathematical calculation on your name to generate a specific hook number.They hang your coat on that exact hook.When you return and give your name, they don't search the room. They do the exact same math, walk straight to that specific hook, and hand you your coat.Because of this direct math, looking up, inserting, or deleting an item in a hashmap takes $O(1)$ time (instantaneous), regardless of whether you have 10 items or 10 million.2. How it Works "Under the Hood"To achieve this magic $O(1)$ speed, a hashmap relies on a Hash Function.When you write my_dict["apple"] = 50, here is what Python does in the background:Hashing: Python passes the string "apple" into a hash function, which scrambles the letters into a massive integer (e.g., 834927492).Compression: Python uses the modulo operator (%) to shrink that massive number to fit inside its underlying memory array (e.g., 834927492 % 8 = 4).Storage: Python stores the value 50 at index 4 of its memory array.

```python
# Creating a Hashmap (Dictionary)
student_grades = {
    "Alice": 95,
    "Bob": 82,
    "Charlie": 88
}
```

```python
# O(1) inserions
student_grades['David'] = 91
```

```python
# O(1) Lookups
student_grades["Alice"]
```

```python
# O(1) Deletion

del student_grades["Bob"]
```

```python
def two_sums(nums, target):
    seen = {}

    for i in range(len(nums)):
        current_num = nums[i]
        complement = target - current_num
        if complement in seen:
            return [seen[complement], i]
        
        seen[complement] = i

        
```
