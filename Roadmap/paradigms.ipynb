1. Procedural Programming (State & Control Flow)

The Philosophy: Procedural programming is all about managing state over time. You set up a variable, write a loop, and modify that variable step-by-step until you reach your goal. It relies heavily on control flow statements (if, elif, while, for) and keeping track of pointers or indexes.

Core Concepts:

Mutability: You are constantly updating and changing existing variables (e.g., count += 1).

Step-by-Step Execution: The computer reads your instructions exactly like a cooking recipe.

```python
nums = [0,1,0,3,12]
for i in nums:
    if i == 0:
        nums.remove(i)
        nums.append(0)





```

2. Object-Oriented Programming (Encapsulation & Memory)

The Philosophy: OOP is about bundling data (state) and behavior (methods) into a single unit called an Object. The biggest advantage of OOP is that an object remembers things between function calls. In procedural code, variables die when the function ends. In OOP, the object stays alive and remembers its state.

Core Concepts:

__init__ (Constructor): This is where the object is born and sets up its initial memory/variables.

self: This is how the object talks to itself. self.count means "my personal count variable".

Encapsulation: Hiding the messy details inside the object so the outside world just presses a simple button (calls a method).

3. Functional Programming (Pure Pipelines)

The Philosophy: Functional Programming treats coding like a factory assembly line. Data goes in on a conveyor belt, passes through machines (functions) that transform it, and comes out the other end.
Crucially, Pure FP never mutates original data. Instead of changing a variable, you always return a brand-new, transformed piece of data.

Core Concepts:

Pure Functions: A function that always returns the exact same output for a given input, and doesn't mess with outside variables.

Higher-Order Functions: Functions that take other functions as arguments (like map(), filter()).

Method Chaining: Combining string or list methods together into a single, elegant pipeline.



```python
s = "Let's take LeetCode contest"
my_list = s.split()
print(reversed(my_list[0]))
# print(my_list[0].reversed())
```

```python
# 1. MaP() function takes a function and a list and apply all the items in a list to the function

def double_number(x):
    return x * 2

numbers = [1, 2, 3, 4, 5]

list(map(double_number, numbers))
```

```python
# 2. filter() function takes a function and return True or False for every item in the list

def is_even(x):
    return x % 2 == 0

list(filter(is_even, numbers))
```

```python
# 3. reduce() takes a list and a function and squash the items in the list into a single value
from functools import reduce

def add_two_numbers(a, b):
    return a + b

numbers = [1, 2, 3, 4]

# Step 1: 1 + 2 = 3
# Step 2: 3 + 3 = 6
# Step 3: 6 + 4 = 10
result = reduce(add_two_numbers, numbers)

print(result)
```

```python
# 4. Zip() takes two or more lists and pairs their elements together side-by-side like a zipper on jacket

names = ["Alice", 'Bob', 'Charlie']
scores = [85, 92, 100]
list(zip(names, scores))
```

```python
# 5. sorted() sorts anything

def get_length(words):
    return len(words)

sorted(names, key=get_length)
```
