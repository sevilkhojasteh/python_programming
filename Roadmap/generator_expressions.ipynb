```python
# List comprehension uses []
# generator expression uses ()

nums = [1, 2, 3, 4, 5]

my_list = [x * 10 for x in nums]

print(my_list)

my_gen = (x * 10 for x in nums)
print(my_gen)
```

```python
my_gen = (x * 10 for x in [1, 2, 3])

print(next(my_gen))
print(next(my_gen))
```

```python
# generators do not store data so they are memory efficient but you can only call them once
# so if you are going to use them multiple times you need to store them in a list

print(list(my_gen))
print(list(my_gen))
```

```python
square_gen = (x**2 for x in range(1, 6))
list(square_gen)
```
