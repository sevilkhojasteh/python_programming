*How data enters and leaves a collection*
1. The stack (LIFO)
- push: Add an item to the top (O(1))
- pop: Remove the item from the top (O(1))
- Real-World Use; The "undo" button

- python built-in list is a perfect example of stack

```python
stack = []

stack.append("A")
stack.append("B")
stack.append("C")

top_item = stack.pop()
print(f"removed, {top_item}")
print(stack)
```

2. The Queue (FIFO)
- Enqueue: Add an item tp the back(O(1))
- Dequeue: Remove an item from the front
- Real-World Use: A printer queue (documents print in the exact order they were sent), or customer service phone systems ("Your call will be answered in the order it was received").

```python
from collections import deque
```

```python
queue = deque()

queue.append("Alice")
queue.append("Bob")
queue.append("Charlie")

first_served = queue.popleft()
print(f"Served: {first_served}") # Output: Served: Alice
print(queue)

queue.pop()
```

**The Heap**


Think of a Heap like a hospital emergency room
A Heap keeps the "most important" item at the very top so it can be accessed instantly ($O(1)$ time).

- Min-Heap: The smallest number is always at the top.
- Max-Heap: The largest number is always at the top.
- Real-World Use: Finding the shortest path in GPS navigation (Dijkstra's Algorithm), or fetching the top 10 highest-scoring players on a leaderboard without sorting the entire millions-long list.

```python
class Solution:
    def is_valid(self, s: str) -> bool:
        stack = []
        bracket_map = {
            ')' : '(',
            '}' : "{",
            ']' : '['
        }

        for char in s:
            if char not in bracket_map:
                stack.append(char)

            else:
                top_element = stack.pop() if stack else '#'

                if bracket_map[char] != top_element:
                    return False
                
        return not stack
```

```python
import heapq

class Solution:
    def findKthLargest(self, nums: list[int], k: int) -> int:
        min_heap = []
        
        for num in nums:
            # Push the current number into the heap
            heapq.heappush(min_heap, num)
            
            # If our VIP list gets too big, kick out the smallest number
            if len(min_heap) > k:
                heapq.heappop(min_heap)
                
        # The root of the min-heap is the Kth largest element
        return min_heap[0]
```
