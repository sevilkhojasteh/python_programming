```python
# There is an error called value error
# for example here if the input is anything but int this error happens

x = int(input("What's is x?"))

print(f"x is {x}")
```

```python
try: 
    x = int(input("What's is x? "))
except ValueError:
    print("x is not an inetegr.")

# Here the error is name error
print(f"x is {x}")
```

```python
try: 
    x = int(input("What's is x? "))
except ValueError:
    print("x is not an inetegr.")
else:
    print(f"x is {x}")
```

```python
while True:
    try: 
        x = int(input("What's is x? "))
    except ValueError:
        print("x is not an inetegr.")
    else:
        break

print(f"x is {x}")
```
