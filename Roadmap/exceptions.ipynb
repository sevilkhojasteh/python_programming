```python
# you can raise when you want to intentionally trigger an exception.
# This usually happens when Python thinks a piece ofdata is perfectly fine, 
# but the rules of your specific program say it is wrong
```

```python
def main():
    x = get_int("What is your age? ")
    print(f"you are {x} years old.")

def get_int(prompt):
    while True:
        try:
            user_input = int(input(prompt))

            if user_input < 0 or user_input > 150:
                raise ValueError("Please enter your age.")
            
            return user_input
        
        except ValueError as e:
            print(f"Invalid input: {e}")


main()
```
