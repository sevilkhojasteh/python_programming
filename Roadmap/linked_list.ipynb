**Array**

an array is a collection of items sorted side-by-side memory locations.
- dynamic array 
- fast lookups (O(1))
- slow insertions (O(n))



```python
# In Python, arrays are just standard lists
my_array = [10, 20, 30, 40]

# O(1) Time: Instant access by index
print(my_array[2]) # Output: 30

# O(n) Time: Inserting at the beginning forces all other elements to shift
my_array.insert(0, 5) 
print(my_array) # Output: [5, 10, 20, 30, 40]
```

**Linked list**
A linked list is a collection of items scattered randomly across memory. To keep track of them, each item (called a Node) stores two things:
1. The actual data
2. A pointer(reference) to the next node in the chain

- Slow Lookups(O(n))
- fast insertions(O(1))


```python
# 1. Create the Node blueprint

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None # The pointer to the next node

# Create the linked list manager

class LinkedList:
    def __init__(self):
        self.head = None # The start of the list

    def display(self):
        current = self.head
        element = []
        while current:
            element.append(str(current.data))
            current = current.next

        print("->".join(element))
```

```python
my_list = LinkedList()

node1 = Node(10)
node2 = Node(20)
node3 = Node(30)
```

```python
my_list.head = node1
node1.next = node2
node2.next = node3

my_list.display()
```
