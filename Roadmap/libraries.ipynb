```python
import requests
import sys
import json

if len(sys.argv) != 2:
    sys.exit()

respond = requests.get("https://itunes.apple.com/search?entity=song&limit=1&term=" + sys.argv[1])

print(json.dumps(respond.json(), indent=2))
```
