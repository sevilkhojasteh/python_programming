```python
# Local

def calculate_daily_cost(kwh_used):
    tarif = 0.15
    total_cost = kwh_used * tarif
    return total_cost
```

```python
print(calculate_daily_cost(45.2))
```

```python
# Global

BASE_VOLTAGE = 220

def calculate_power(amps):
    return BASE_VOLTAGE * amps


print(calculate_power(10))
```

```python
# Modifying a global Varibale

total_kwh_saved = 0

def record_savings(kws):
    global total_kwh_saved
    total_kwh_saved += kws
    return total_kwh_saved
```

```python
record_savings(12.5)
```

```python
record_savings(8.0)
```

```python
# Enclosing scope

def setup_sensor(sensor_id):
    status = 'Active'

    def ping_sensor():
        print(f"Sensor {sensor_id} is {status}")

    ping_sensor()
```

```python
setup_sensor("Meter-A1")
```

```python
# Modifying an enclosing variable

def smart_thermostat():
    tempretature = 20 #enclosing variable

    def increase_temp():
        nonlocal tempretature # Allows us to modify the enclosing variable
        tempretature += 1
        return tempretature
    
    print(increase_temp())
    print(increase_temp())
    print(increase_temp())
```

```python
smart_thermostat()
```

```python
# Exercise

# 1. GLOBAL Scope
total_power_grid = 5000

def house_manager():
    # 2. ENCLOSING Scope
    house_power = 0
    min_value = 50  # Just a random variable tracking minimum threshold
    
    def plug_in_appliance(watts):
        # 3. LOCAL Scope
        status = f"Plugging in appliance drawing {watts}W..."
        print(status)
        
        # ERROR 1: Trying to modify an ENCLOSING variable
        nonlocal house_power
        house_power += watts
        
        # ERROR 2: Trying to modify a GLOBAL variable
        global total_power_grid
        total_power_grid -= watts
        
    # Running the inner function twice
    plug_in_appliance(1000)
    plug_in_appliance(500)
    
    # ERROR 3: Built-in Scope Trap
    past_usage = [400, 800, 1200]
    lowest_usage = min(past_usage) # Trying to use the built-in min() function
    
    return house_power, lowest_usage

# Execute the system
final_house, final_lowest = house_manager()
print(f"House Total: {final_house}W")
print(f"Grid Remaining: {total_power_grid}W")
```
