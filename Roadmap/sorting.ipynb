```python
import time
import random

n = 1000
li = random.sample(range(1, 1001), n)
```

1. Slow sorting algorithms
- Bubble sort
- Insertion sort


```python
# Bubble sort
def bubble_sort(arr):
    n = len(arr)

    for i in range(n):
        for j in range(0, n-i-1):
            if arr[j] > arr[j + 1]:
                arr[j] , arr[j+1] = arr[j+1], arr[j]

    return arr
```

```python
start = time.time()
bubble_sort(li)
end = time.time()

print(f"{end - start:.4f}")
```

```python
# Insertion sort 
def insertion_sort(arr):
    for i in range(1, len(arr)):

        key = arr[i]
        j = i - 1
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr
```

```python
start = time.time()
insertion_sort(li)
end = time.time()

print(f"{end - start:.4f}")
```

```python
def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr) // 2
        left_half = arr[:mid]
        right_half = arr[mid:]

        merge_sort(left_half)
        merge_sort(right_half)

        # Merge the two halves back together
        i = j = k = 0
        while i < len(left_half) and j < len(right_half):
            if left_half[i] < right_half[j]:
                arr[k] = left_half[i]
                i += 1
            else:
                arr[k] = right_half[j]
                j += 1
            k += 1

        # Catch any remaining elements
        while i < len(left_half):
            arr[k] = left_half[i]
            i += 1
            k += 1
        while j < len(right_half):
            arr[k] = right_half[j]
            j += 1
            k += 1
    return arr
```

```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    else:
        # Pick the last element as the pivot
        pivot = arr.pop()
        
        # Elements less than the pivot go left, greater go right
        greater = [x for x in arr if x > pivot]
        lesser = [x for x in arr if x <= pivot]
        
        # Recursively sort and combine
        return quick_sort(lesser) + [pivot] + quick_sort(greater)
```
