\d : Any digit (0-9).

\D : Any non-digit.

\w : Any word character (a-z, A-Z, 0-9, and underscore _).

\W : Any non-word character (spaces, punctuation).

\s : Any whitespace (space, tab, newline).

. : Any character whatsoever (except a newline).


+ : One or more of the previous character. (e.g., \d+ finds "1", "12", "123").

* : Zero or more of the previous character.

? : Zero or one of the previous character (makes it optional).

{n} : Exactly n times. (e.g., \d{3} finds exactly 3 digits).

{n,m} : Between n and m times.

^ : The start of a string.

$ : The end of a string.

\b : A word boundary (ensures you don't match "cat" inside "category").


[abc] : Matches exactly one 'a', 'b', or 'c'.

[a-z] : Matches any lowercase letter.

[^a-z] : The ^ inside brackets means NOT. Matches anything except a lowercase letter.

(abc) : A capture group. Groups multiple tokens together and "saves" the exact match for later extraction.

a|b : Logical OR. Matches 'a' or 'b'.

```python
import re

text = "My number are 555-1234 and 555-9876"

pattern = r"\d{3}-\d{4}"

matches = re.findall(pattern, text)

matches
```

```python
text = "User email is admin@system.com today."

pattern = r"\w+@\w+\.\w+"

match = re.findall(pattern, text)

match
```

```python
text = "The cost is $100 and $50"

clean_text = re.sub(r"\$\d+", "[REDACTED]", text)

clean_text
```

```python
text = "I went to the the store to buy some some milk."
duplicates = [match.group() for match in re.finditer(r"\b(\w+)\s+\1\b", text)]

duplicates
```

```python
keys = ["sk_test_123456", "sk_live_987654", "sk_demo_123456", "sk_test_1234"]

pattern = r"sk_(test|live)_\d{6}"

for key in keys:
    if re.search(pattern, key):
        print(f"Valid: {key}")
    else:
        print(f"Invalid: {key}")
```

```python
import re

olympics = "1920 Antserp olympics, 1924 Paris olympics, 1928 Amsterdam olympics"

# Pattern: Capture 4 digits (Year), a space, and capture a word (City)
pattern = r"(\d{4})\s([A-Za-z]+)"

# findall with groups returns a list of tuples: [('1920', 'Antserp'), ...]
matches = re.findall(pattern, olympics)

# Separate them using a quick list comprehension
years = [match[0] for match in matches]
cities = [match[1] for match in matches]

print("Years:", years)
print("Cities:", cities)
```
