```python
names = ["Alice", "Bob", "Charlie"]
names[0]  # Output: 'Alice'
names[1]  # Output: 'Bob'
names[2]  # Output: 'Charlie'
names[-1]  # Output: 'Charlie'
names[-2]  # Output: 'Bob'
names[-3]  # Output: 'Alice'
```

```python
names[1:3]  # Output: ['Bob', 'Charlie']
```

```python
names[0] = "Dave"
names  # Output: ['Dave', 'Bob', 'Charlie']
```

```python
names.append("Eve")
names  # Output: ['Dave', 'Bob', 'Charlie', 'Eve']
```

```python
names.insert(1, "Frank")
print(names)  # Output: ['Dave', 'Frank', 'Bob', 'Charlie', 'Eve']
names.remove("Bob")
print(names)  # Output: ['Dave', 'Frank', 'Charlie', 'Eve']
names.pop()  # Output: 'Eve'
print(names)  # Output: ['Dave', 'Frank', 'Charlie']
names.pop(1)  # Output: 'Frank'
print(names)  # Output: ['Dave', 'Charlie']
```

```python
del names[0]
print(names)  # Output: ['Charlie']
```

```python
names.clear()
print(names)  # Output: []
```
