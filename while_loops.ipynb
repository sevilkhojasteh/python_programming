```python
counter = 0
while counter < 5:
    print("The count is at", counter)
    counter += 1
```

```python
# Nested while loop example
outer_counter = 0
while outer_counter < 3:
    inner_counter = 0
    while inner_counter < 2:
        print(f"Outer counter: {outer_counter}, Inner counter: {inner_counter}")
        inner_counter += 1
    outer_counter += 1
```
