```python
# for loop is called definite iteration
names = [
    'a',
    'b',
    'c',
    'd'
]
```

```python
for name in names:
    print(name)
```

```python
person = {
    'a': 1,
    'b': 2,
    'c': 3
}
```

```python
for key in person:
    print(key)
```

```python
for value in person.values():
    print(value)
```

```python
for item in person.items():
    key, value = item
    print(item)
    print(key)
    print(value)
```

```python
# The enumerate() function

names = [
    'a',
    'b',
    'c',
    'd',
    'e'
]
```

```python
for index, name in enumerate(names, start=1):
    print(index, name)
```

```python
# Example of a nested loop
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
```

```python
# Iterate over each row in the matrix
for row in matrix:
    # Iterate over each element in the row
    for element in row:
        print(element)
```
