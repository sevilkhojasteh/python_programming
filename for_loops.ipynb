```python
# for loop is called definite iteration
names = [
    'a',
    'b',
    'c',
    'd'
]
```

```python
for name in names:
    print(name)
```

```python
person = {
    'a': 1,
    'b': 2,
    'c': 3
}
```

```python
for key in person:
    print(key)
```

```python
for value in person.values():
    print(value)
```

```python
for item in person.items():
    key, value = item
    print(item)
    print(key)
    print(value)
```
