```python
# In override

class Vehicle:
    def display_info(self):
        print("this is a vehicle")

class Car(Vehicle):
    def display_info(self):
        print("This is a car")

class Motorcycle(Vehicle):
    def display_info(self):
        print("This is a motorcycle")
```

```python
def vehicle_info(vehicle):
    vehicle.display_info()
```

```python
car = Car()
motor = Motorcycle()
```

```python
vehicle_info(car)
```

```python
vehicle_info(motor)
```

```python
class Animal:
    def speak(self):
        print("This animal does not have a sound")
    
class Dog(Animal):
    def speak(self):
        print("Woof!")

class Cat(Animal):
    def speak(self):
        print("Meow!")
```

```python
dog = Dog()
cat = Cat()

def animal_speak(animal):
    animal.speak()

```

```python
animal_speak(cat)
```

```python
animals = [Dog(), Cat()]

for animal in animals:
    animal.speak()
```
