```python
# Instance methods

class Dog:
    def __init__(self, name, age):
        # Each Dog instances has attributes like name, age
        self.name = name
        self.age = age
    # here bark is an instance method
    def bark(self):
        return f"{self.name} says woof!"
```

```python
my_dog = Dog("Buddy", 4)
```

```python
my_dog.bark()
```

```python
# Static method
"""Note that parameters do not include self. This is the key distinction between static and instance methods. 
    Static methods are not tied to any instance or class state, and they are not able to access or modify instance or class attributes directly.

Static methods shine in scenarios where you need to perform a function that's related to 
the class but doesn't necessarily interact with class-specific data.
"""
```

```python
class MathOperations:
    @staticmethod
    def calculate_square_root(number):
        return number ** 0.5
```

```python
result = MathOperations.calculate_square_root(9)
```

```python
result
```

```python
class TemperatureConverter:
    @staticmethod
    def celsius_to_fahrenheit(celsius):
        return (celsius * 9/5) + 32
```

```python
celsius_temp = 25
fahrenheit_temp = TemperatureConverter.celsius_to_fahrenheit(celsius_temp)
f"{celsius_temp}°C is equal to {fahrenheit_temp}°F"
```

```python
# Class method:
"""
This signals to Python that the following method is not just any 
method but one that should automatically receive
the class itself as the first argument instead of an instance. 
This first parameter is conventionally named cls
"""
```

```python
class Inventory:
    discount_rate = 10
    def __init__(self, name, price):
        self.name = name
        self.price = price

    @classmethod
    def set_discount_rate(cls, new_rate):
        cls.discount_rate = new_rate
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    @classmethod
    def from_string(cls, employee_str):
        name, salary = employee_str.split('-')
        return cls(name, int(salary))
emp = Employee.from_string("John-5000")
emp.name, emp.salary
```
