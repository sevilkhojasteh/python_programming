```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Student(Person):
    def __init__(self, name, age, student_id):

        super().__init__(name, age)
        self.student_id = student_id
    def info(self):
        print(self.name, self.age, self.student_id)
```

```python
obj2 = Student("sevil", 24, "1234")

obj2.info()

```

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __str__(self):
        return f"'{self.title}' by {self.author}"
    
    def __repr__(self):
        return f"Book({self.title!r}, {self.author!r})"
```

```python
obj = Book("a", "b")
```

```python
print(obj)
```

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __str__(self):
        return f"'{self.title}' by {self.author}"
    
    def __repr__(self):
        return f"{self.__class__.__name__}(title={self.title}, author={self.author})"


# How to make repr more dynamic
```

```python
obj = Book("a", "b")
```

```python
obj
```

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
```

```python
obj1 = Point(3, 4)
obj2 = Point(6, 8)
obj3 = Point(3, 4)
```

```python
obj1 == obj2
```

```python
obj1 == obj3
```

```python
obj1.__eq__(obj3)
```
