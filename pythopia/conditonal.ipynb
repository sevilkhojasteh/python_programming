```python
class car:
    def __init__(self, speed):
        self.speed = speed

    def __bool__(self):
        return self.speed > 0
```

```python
obj = car(10)
```

```python
bool(obj)
```

```python
obj = car(0)
```

```python
bool(obj)
```
