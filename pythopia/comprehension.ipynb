```python
l2 = []
for num in l1:
    l2.append(num ** 2)
    
print(l2)
```

```python
names = [
    'ali', 
    'hasan', 
    'rasa', 
    'kian'
]
```

```python
{name: len(name) for name in names}
```

```python
[i for i in range(10)]
```

```python
{name: len(name) for name in names if name != 'hasan'}
```

```python
d = {}
for name in names:
    if name != 'hasan':
        d[name] = len(name)
```
