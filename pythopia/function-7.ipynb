```python
square = lambda num: num * num
```

```python
square(4)
```

 **Lambda function**

 Just in one line
 you cannot have a two-line lambda function

 *When to use*

 short-lived functions

 higher-order functions --> Map function

 Small transformation and actions
 
 functional constructs

```python
list(map(square, [1, 2, 3, 4, 5]))
```

```python
list(map(lambda x: x * 3, [1, 2, 3, 4, 5]))
```

```python
list(filter(lambda x: x % 2 ==0, [1, 2, 3, 4, 5]))
```

```python
list(filter(lambda x: x % 2 !=0, [1, 2, 3, 4, 5]))
```

```python
sorted(
    [(1, 2), (3, 4), (5, 8), (-1, 2)],
    key = lambda x:x[0]
)
```

```python
f = lambda *args: sum(args)
```

```python
f(1, 2, 3, 4, 5, 6)
```
