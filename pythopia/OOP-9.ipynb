```python
# Encapsulation
# some features we don't want the use to see --> decreasing the complexity

# These techniques are not forced
# in c++ : private, protected, public
# in python private is name normal
# protected: _name
# private: __name
```

```python
# avoiding unintended modifying
# Data hiding, Data protection, Abstraction
```

```python
class Vehicle:
    def __init__(self, make, model):
        # These are public attributes
        # We have access from inside of the class and outside
        self.make = make
        self.model = model

    def display_info(self):
        print(f"{self.name} {self.model}")
        
```

```python
obj = Vehicle("toyota", "Land cruise")
```

```python
obj.make, obj.model
```

```python
class Vehicle:
    def __init__(self, make, model, fuel_capacity):
        self.make = make
        self.model = model
        self._fuel_capacity = fuel_capacity
        self._fuel_level = fuel_capacity

    def _consume_fuel(self, amount):
        if self._fuel_capacity - amount < 0:
            print("Not enough fuel.")

        else: 
            self._fuel_level -= amount

        print(f"Fuel level: {self._fuel_level}")
```

```python
obj = Vehicle("toyota", "Land Cruise", 50)
```

```python
class Sedan(Vehicle):
    def drive(self, distance):
        self._consume_fuel(distance * 0.1)
```

```python
class SUV(Vehicle):
    def drive(self, distance):
        self._consume_fuel(distance * 0.15)
```

```python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner             # Public attribute
        self.__balance = balance       # Private attribute

    def __update_balance(self, amount):  # Private method
        self.__balance += amount
        return self.__balance

    def deposit(self, amount):
        if amount > 0:
            new_balance = self.__update_balance(amount)
            print(f"Deposit successful. New balance: {new_balance}")
        else:
            print("Deposit amount must be positive.")
```
