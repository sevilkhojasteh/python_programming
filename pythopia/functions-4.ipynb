```python
def find_book(title):
    books = ['a', 'b', 'c', 'd']
    if title in books:
        return True
    else: return False
```

```python
book_exist = find_book('a')
```

```python
def process_number(number):
    if type(number) != int:
        return f'{number} is not a number'
    if number % 2 == 0:
        return f'{number} is even'
    else: return f'{number} is odd'
```

```python
process_number(4)
```
