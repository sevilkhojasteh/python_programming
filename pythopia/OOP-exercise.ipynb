```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_checked_out = False

    def toggle_status(self):
        self.is_checked_out = not self.is_checked_out
        status = "checked out" if self.is_checked_out else "available"
        print(f"{self.title} is now {status}")

    def __str__(self):
        status = "checked out" if self.is_checked_out else "available"
        return f"Book: {self.title} | Author: {self.author} | Status: {status}"
```

```python
book1 = Book("wuthering heights", "Emily Bronte")
book2 = Book("Pride and prejedious", "jane ustin")
```

```python
print(book1)
```

```python
book1.toggle_status()
```

```python
class SmartLight:
    def __init__(self, room_name):
        self.room_name = room_name
        self.is_on = False

    def turn_on(self):
        self.is_on = True

    def turn_off(self):
        self.is_on = False

    def get_status(self):
        status = "ON" if self.is_on else "OFF"
        print(f"The {self.room_name} light is {status}")
        
```

```python
kitchen = SmartLight("kitchen")
bed_room = SmartLight("bed room")
```

```python
bed_room.turn_on()
```

```python
bed_room.get_status()
```
