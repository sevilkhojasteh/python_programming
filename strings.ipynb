```python
# What is a string?
# A string is a sequence of characters. It can be used to represent text, numbers, or any other type of data. In Python, strings are enclosed in either single quotes (' ') or double quotes (" ").
# Creating a string
my_string = "Hello, World!"
print(my_string)  # Output: Hello, World!
# String concatenation
greeting = "Hello"
name = "Alice"
full_greeting = greeting + ", " + name + "!"
print(full_greeting)  # Output: Hello, Alice!
# String repetition
repeated_string = "Ha" * 3
print(repeated_string)  # Output: HaHaHa
# String indexing
first_character = my_string[0]
print(first_character)  # Output: H
# String slicing
substring = my_string[0:5]
print(substring)  # Output: Hello
# String methods
upper_string = my_string.upper()
print(upper_string)  # Output: HELLO, WORLD!
lower_string = my_string.lower()
print(lower_string)  # Output: hello, world!
# String length
string_length = len(my_string)
print(string_length)  # Output: 13
# String formatting
name = "Bob"
age = 30
formatted_string = f"My name is {name} and I am {age} years old."
print(formatted_string)  # Output: My name is Bob and I am 30 years

# String immutability
# In Python, strings are immutable, which means that once a string is created, it cannotbe changed. However, you can create a new string by concatenating or slicing existing strings.
original_string = "Hello"
new_string = original_string + " World"
print(new_string)  # Output: Hello World       
```

```python
# String formating:
# the .format() method allows you to insert values into a string using placeholders. You can use curly braces {} as placeholders and pass the values to be inserted as arguments to the format() method.
greeting = "Hello, {}"
name = "Alice"
greeting.format(name)  # Output: 'Hello, Alice'
```

```python
greeting = "Hello, {name}"
greeting.format(name="Alice")  # Output: 'Hello, Alice'
# instead of position-based formatting, you can also use named placeholders in the string and pass the values as keyword arguments to the format() method. This can make your code more readable and easier to maintain, especially when you have multiple placeholders in the string.


```

```python
# f-strings: f-strings (formatted string literals) are a more concise and readable way to format strings in Python. They allow you to embed expressions inside string literals, using curly braces {}. You can prefix the string with the letter 'f' to indicate that it is an f-string.
name = "Alice"
age = 30
f"My name is {name} and I am {age} years old."  # Output: 'My name is Alice and I am 30 years old.'
```
