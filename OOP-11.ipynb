```python
class Employee:
    def work(self):
        print("I am working")
        print("I am an emplyee")

class SoftWareEngineer(Employee):
    def work(self):
        super().work()
        print("Writing code and fixing bugs.")
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Student(Person):
    def __init__(self, name, age, student_id):

        super().__init__(name, age)
        self.student_id = student_id
    def info(self):
        print(self.name, self.age, self.student_id)
```

```python
obj2 = Student("sevil", 24, "1234")

obj2.info()

```

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __str__(self):
        return f"'{self.title}' by {self.author}"
    
    def __repr__(self):
        return f"Book({self.title!r}, {self.author!r})"
```

```python
obj = Book("a", "b")
```

```python
print(obj)
```
