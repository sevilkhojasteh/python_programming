```python
# Argument Passing
def create_invitation(name, age, date):
    return(f"{name} is {age} years old invited to party on {date}")
```

```python
create_invitation("john", 25, "2026-12-30")
```

```python
create_invitation("john", 25)
```

```python
# This is keyword argument
create_invitation(name="John", age=25, date="2026")
```

```python
# you can combine positional and keyword arguments but if you start the keyword you need to do the rest keyword too
create_invitation("John", age=25, date="2026")
```

```python
# you can put default value

def create_profile(name, age, country='USA', language='English'):
    profile = {
        'name': name,
        'age': age,
        'country': country,
        'language': language
    }
    return profile
```

```python
create_profile("john", 25)
```

```python
create_profile("sevil", 24, "Iran")
```
